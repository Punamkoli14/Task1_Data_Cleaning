```python
missing_values=data.isnull().sum()
```

```python
missing_values
```

```python
data.duplicated().value_counts()
```

```python
data_new= data.drop_duplicates()
```

```python
data_new
```

```python
data_types = data_new.dtypes
```

```python
data_types
```

```python
data_new.columns = data_new.columns.str.strip().str.lower().str.replace(' ', '_')
data_new['country'] = data_new['country'].str.strip().str.title()
```

```python
{
    "Missing Values": missing_values,
    "Data Types": data_types,
    "Shape Before Cleaning": data.shape,
    "Shape After Removing Duplicates": data_new.shape,
    "Cleaned Column Names": data_new.columns.tolist()
}
```

```python
data_new = data_new.copy()
data_new['date'] = pd.to_datetime(data_new['date'], errors='coerce')
data_new['date'] = data_new['date'].dt.strftime('%d-%m-%Y')
```
